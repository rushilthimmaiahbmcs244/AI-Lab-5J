```python
import random

# Fitness function
# We want to maximize f(x) = -(x - 5)^2 + 25
# Maximum value is 25 when x = 5
def fitness(x):
    return -(x - 5) ** 2 + 25


# Parameters
POPULATION_SIZE = 10
GENERATIONS = 50
CROSSOVER_RATE = 0.8
MUTATION_RATE = 0.1

MIN_VALUE = 0
MAX_VALUE = 10


# Generate initial population
def create_population():
    population = []

    for i in range(POPULATION_SIZE):
        x = random.uniform(MIN_VALUE, MAX_VALUE)
        population.append(x)

    return population


# Select the best individuals
def selection(population):
    population.sort(key=fitness, reverse=True)

    # Select top half
    return population[:POPULATION_SIZE // 2]


# Crossover
def crossover(parent1, parent2):
    if random.random() < CROSSOVER_RATE:
        alpha = random.random()

        child1 = alpha * parent1 + (1 - alpha) * parent2
        child2 = alpha * parent2 + (1 - alpha) * parent1

        return child1, child2

    return parent1, parent2


# Mutation
def mutation(x):
    if random.random() < MUTATION_RATE:
        x = x + random.uniform(-1, 1)

    # Keep value within limits
    x = max(MIN_VALUE, min(MAX_VALUE, x))

    return x


# Genetic Algorithm
population = create_population()

for generation in range(GENERATIONS):

    # Evaluate and select parents
    parents = selection(population)

    new_population = []

    # Create new population
    while len(new_population) < POPULATION_SIZE:

        parent1 = random.choice(parents)
        parent2 = random.choice(parents)

        # Crossover
        child1, child2 = crossover(parent1, parent2)

        # Mutation
        child1 = mutation(child1)
        child2 = mutation(child2)

        new_population.append(child1)

        if len(new_population) < POPULATION_SIZE:
            new_population.append(child2)

    population = new_population


# Find best solution
best_solution = max(population, key=fitness)

print("Best solution:", best_solution)
print("Best fitness:", fitness(best_solution))
```
